# Small KSAT demo

This notebook runs the public random 3-SAT constructive accessibility demo and then computes
the gap profile from the illustrative example curves. See the README and the
[paper](https://doi.org/10.5281/zenodo.19225548) for context.


```
from pathlib import Path
from reachability_labs_demo.ksat_adapter import KSatDemoAdapter
from reachability_labs_demo.runtime import SweepConfig, run_sweep
from reachability_labs_demo.plotting import plot_success_curve, plot_hazard_curve, plot_variance_decomposition
out = Path('examples/notebook_demo')
adapter = KSatDemoAdapter()
cfg = SweepConfig(n=100, alphas=[1.0,1.5,2.0,2.5,3.0,3.5,4.0,4.5,5.0], trials=24, branches_per_instance=3, exact_local=True, base_seed=1)
_, summary, hazard, variance = run_sweep(adapter, cfg, out)
plot_success_curve(summary, out/'success_curve.png')
alpha_focus = min(summary, key=lambda r: abs(float(r['success_rate'])-0.5))['alpha']
plot_hazard_curve(hazard, out/'hazard_curve.png', float(alpha_focus))
plot_variance_decomposition(variance, out/'variance_decomposition.png')
summary[:3]
```

```
from reachability_labs_demo.gap_profile import compute_gap_profile, load_curves_csv
from reachability_labs_demo.plotting import plot_gap_profile
alpha, existence, reachability = load_curves_csv('../examples/example_curves.csv')
profile = compute_gap_profile(alpha, existence, reachability)
profile.midpoint_gap, profile.peak_gap, profile.peak_level
```
